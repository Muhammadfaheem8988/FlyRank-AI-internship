## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# 1. Load the starter dataset
df = pd.read_csv("/content/content_refresh_anonymized.csv")

# 2. Compute 3 core evidence numbers
total_pages = len(df)
declining_pages = (df["trend_direction"] == "down").sum()
declining_pct = (declining_pages / total_pages) * 100

high_traffic_decay = df[(df["trend_direction"] == "down") & (df["impressions_90d"] >= 500)].shape[0]

print(f"Total pages in starter slice: {total_pages:,}")
print(f"Pages trending down: {declining_pages:,} ({declining_pct:.1f}%)")
print(f"High-visibility decaying pages (Impressions >= 500 & Trend = down): {high_traffic_decay:,}")
```

**Evidence from Starter Data:**
1. There are **30,000 total content items** in the dataset.
2. A substantial portion of pages show declining performance (`trend_direction == 'down'`).
3. Out of all declining pages, a critical subset has high visibility ($\ge 500$ impressions in 90 days), proving that prioritization is necessary so editors don't waste time on zero-traffic pages.
